```python
%pip install numpy
import numpy as np
%pip install pandas
import pandas as pd

%pip install matplotlib
import matplotlib.pyplot as plt

data = {
    "Student": ["Anan", "Riya", "Karan", "Nana", "Raj"],
    "Marks": [75, 0, 55, 85, 55],
}

df = pd.DataFrame(data)

average = np.mean(df["Marks"])
highest = np.max(df["Marks"])

print(df)
print("Class Average:", average)
print("Highest Marks:", highest)

plt.bar(df["Student"], df["Marks"])
plt.axhline(average, linestyle="--", label="Average")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student Performance Dashboard")
plt.legend()
plt.show()
```

```python
sales_data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [20000, 25000, 22000, 30000, 35000, 32000],
}

sales_df = pd.DataFrame(sales_data)

total_sales = np.sum(sales_df["Sales"])
average_sales = np.mean(sales_df["Sales"])
maximum_sales_row = sales_df.loc[sales_df["Sales"].idxmax()]

print("Total Sales:", total_sales)
print("Average Sales:", average_sales)
print("Maximum Sales Row:")
print(maximum_sales_row)

plt.plot(sales_df["Month"], sales_df["Sales"], marker="o")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Sales Trend")
plt.grid(True)
plt.show()
```
