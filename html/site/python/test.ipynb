```python
%pip install numpy
import numpy as np
%pip install pandas
import pandas as pd

%pip install matplotlib
import matplotlib.pyplot as plt

data = {
    "Student": ["Anan", "Riya", "Karan", "Nana", "Raj"],
    "Marks": [75, 0, 55, 85, 55],
}

df = pd.DataFrame(data)

average = np.mean(df["Marks"])
highest = np.max(df["Marks"])

print(df)
print("Class Average:", average)
print("Highest Marks:", highest)

plt.bar(df["Student"], df["Marks"])
plt.axhline(average, linestyle="--", label="Average")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student Performance Dashboard")
plt.legend()
plt.show()
```

```python
sales_data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [20000, 25000, 22000, 30000, 35000, 32000],
}

sales_df = pd.DataFrame(sales_data)

total_sales = np.sum(sales_df["Sales"])
average_sales = np.mean(sales_df["Sales"])
maximum_sales_row = sales_df.loc[sales_df["Sales"].idxmax()]

print("Total Sales:", total_sales)
print("Average Sales:", average_sales)
print("Maximum Sales Row:")
print(maximum_sales_row)

plt.plot(sales_df["Month"], sales_df["Sales"], marker="o")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Sales Trend")
plt.grid(True)
plt.show()
```

```python
data = {
    "Student": ["Aman", "Riya", "Karr", "Neha"],
    "Python": [78, 92, 65, 88],
    "Maths": [72, 85, 70, 80],
    "DBMS": [80, 89, 60, 84],
}

df = pd.DataFrame(data)

subjects = ["Python", "Maths", "DBMS"]
df["Total"] = df[subjects].sum(axis=1)
df["Average"] = df[subjects].mean(axis=1)

print(df)

plt.plot(df["Student"], df["Python"], marker="o", label="Python")
plt.plot(df["Student"], df["Maths"], marker="o", label="Maths")
plt.plot(df["Student"], df["DBMS"], marker="o", label="DBMS")

plt.xlabel("Student")
plt.ylabel("Marks")
plt.title("Subject-wise Performance")
plt.legend()
plt.grid(True)
plt.show()
```

```python
data = {
    "Employee": ["A", "B", "C", "D"],
    "Department": ["IT", "HR", "IT", "Sales"],
    "Salary": [150000, 40000, 60000, 45000],
}

df = pd.DataFrame(data)

average = df["Salary"].mean()
above_average = df[df["Salary"] > average]
dept_salary = df.groupby("Department")["Salary"].mean()

print(df)
print("Average Salary:", average)
print("Employees earning above average:")
print(above_average)

dept_salary.plot(kind="bar")
plt.ylabel("Average Salary")
plt.title("Average Salary by Department")
plt.show()
```

```python
data = {
    "Day": ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"],
    "Temperature": [30, 32, 35, 29, 36, 34, 31],
}

df = pd.DataFrame(data)

average = np.mean(df["Temperature"])
hot_days = df[df["Temperature"] > average]

print("Average Temperature:", average)
print("Hot Days:")
print(hot_days)

plt.plot(df["Day"], df["Temperature"], marker="o", label="Temperature")
plt.axhline(average, linestyle="--", label="Average")
plt.xlabel("Day")
plt.ylabel("Temperature")
plt.title("Weekly Weather Dashboard")
plt.legend()
plt.grid(True)
plt.show()
```

```python
data = {
    "Student": ["Aman", "Riya", "Karan", "Neha", "Raj", "Sara"],
    "CGPA": [8.2, 9.1, 7.0, 8.8, 6.5, 9.0],
    "Placed": ["Yes", "Yes", "No", "Yes", "No", "Yes"],
    "Package": [6.0, 10.0, 0.0, 8.0, 0.0, 9.0],
}

df = pd.DataFrame(data)

placed = df[df["Placed"] == "Yes"]
average_package = placed["Package"].mean()

print("Average Package:", average_package)
print(placed)

plt.scatter(df["CGPA"], df["Package"])
plt.xlabel("CGPA")
plt.ylabel("Package (LPA)")
plt.title("CGPA vs Placement Package")
plt.grid(True)
plt.show()
```
